```python
# Split data chronologically: 80% train, 20% test (time series)
SPLIT_RATIO = 0.8
split_idx = int(len(df) * SPLIT_RATIO)
SPLIT_DATE = df.loc[split_idx, 'DateTime']

train_df = df.iloc[:split_idx].reset_index(drop=True)
test_df  = df.iloc[split_idx:].reset_index(drop=True)

print(f"Split date: {SPLIT_DATE}")
print(f"Train: {len(train_df):,} rows ({len(train_df)/len(df)*100:.1f}%)  ({train_df['DateTime'].min()} -> {train_df['DateTime'].max()})")
print(f"Test:  {len(test_df):,} rows ({len(test_df)/len(df)*100:.1f}%)  ({test_df['DateTime'].min()} -> {test_df['DateTime'].max()})")

def evaluate_forecast(y_true, y_pred, model_name):
    mae  = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mape = np.mean(np.abs((np.array(y_true) - np.array(y_pred)) / np.array(y_true))) * 100
    print(f"{model_name:15s} | MAE: {mae:8.2f} MW | RMSE: {rmse:8.2f} MW | MAPE: {mape:6.2f}%")
    return {"Model": model_name, "MAE": mae, "RMSE": rmse, "MAPE": mape}

results = []
```

```python
# Define a reusable time-series cross-validation runner, shared across ARIMA, Random Forest, and LSTM (defines the evaluation framework)
from sklearn.model_selection import TimeSeriesSplit

def run_timeseries_cv(data, fold_fn, label, n_splits=5):
    tscv = TimeSeriesSplit(n_splits=n_splits)
    cv_results = []

    for fold, (train_idx, test_idx) in enumerate(tscv.split(data), start=1):
        train_fold = data.iloc[train_idx].reset_index(drop=True)
        test_fold  = data.iloc[test_idx].reset_index(drop=True)

        y_true, y_pred = fold_fn(train_fold, test_fold)

        mae  = mean_absolute_error(y_true, y_pred)
        rmse = np.sqrt(mean_squared_error(y_true, y_pred))
        mape = np.mean(np.abs((np.array(y_true) - np.array(y_pred)) / np.array(y_true))) * 100

        print(f"[{label}] Fold {fold}: Train {len(train_fold):,} | Test {len(test_fold):,} "
              f"| MAE {mae:.2f} MW | RMSE {rmse:.2f} MW | MAPE {mape:.2f}%")

        cv_results.append({"Model": label, "Fold": fold, "MAE": mae, "RMSE": rmse, "MAPE": mape})

    cv_df = pd.DataFrame(cv_results)
    print(f"\n=== {label} CV Summary ===")
    print(f"MAE  mean: {cv_df['MAE'].mean():.2f}  std: {cv_df['MAE'].std():.2f}")
    print(f"RMSE mean: {cv_df['RMSE'].mean():.2f}  std: {cv_df['RMSE'].std():.2f}")
    print(f"MAPE mean: {cv_df['MAPE'].mean():.2f}%  std: {cv_df['MAPE'].std():.2f}%")
    return cv_df
```

```python
# Run the Augmented Dickey-Fuller test to check stationarity
from statsmodels.tsa.stattools import adfuller

series_full = train_df.set_index('DateTime')['SYSTEM_DEMAND_MW']

adf_result = adfuller(series_full)
print(f"ADF Statistic: {adf_result[0]:.4f}")
print(f"p-value:       {adf_result[1]:.4f}")
if adf_result[1] < 0.05:
    print("=> Series is stationary (reject H0).")
else:
    print("=> Series is NOT stationary — differencing will be applied by auto_arima (d>0).")
```

```python
# Use auto_arima to search for the best (p,d,q) order
import pmdarima as pm

ARIMA_WINDOW_HOURS = 24 * 365 * 2
arima_train_series = series_full.iloc[-ARIMA_WINDOW_HOURS:]

print(f"Selecting ARIMA order on {len(arima_train_series):,} recent training hours...")

auto_model = pm.auto_arima(
    arima_train_series,
    start_p=0, start_q=0, max_p=5, max_q=5,
    d=None,
    seasonal=False,
    trace=True,
    error_action='ignore',
    suppress_warnings=True,
    stepwise=True
)

print(auto_model.summary())
order = auto_model.order
print(f"\nSelected order (p,d,q): {order}")
```

```python
# Fit the final ARIMA model and forecast the test period blind (multi-step, no peeking)
from statsmodels.tsa.arima.model import ARIMA

print("Fitting final ARIMA model — this may take a few minutes...")
arima_final = ARIMA(arima_train_series, order=order)
arima_fit = arima_final.fit()

n_forecast = len(test_df)
arima_forecast = arima_fit.forecast(steps=n_forecast)

test_df['ARIMA_Pred'] = arima_forecast.values
results.append(evaluate_forecast(test_df['SYSTEM_DEMAND_MW'], test_df['ARIMA_Pred'], "ARIMA"))
```

```python
# Plot ARIMA's forecast against actual demand for the first two weeks of the test set
plot_slice = test_df.iloc[:24*14]

plt.figure(figsize=(14, 4))
plt.plot(plot_slice['DateTime'], plot_slice['SYSTEM_DEMAND_MW'], label='Actual', color='black', linewidth=1)
plt.plot(plot_slice['DateTime'], plot_slice['ARIMA_Pred'], label='ARIMA Forecast', color='#B85042', linewidth=1)
plt.title('ARIMA — Actual vs Predicted (First 2 Weeks of Test Set)')
plt.xlabel('Date'); plt.ylabel('Demand (MW)'); plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Generate ARIMA walk-forward, one-step-ahead predictions using the true previous value each step
# Build one continuous series across train + test
full_series = df.set_index('DateTime')['SYSTEM_DEMAND_MW']

n_train = len(train_df)
n_total = len(full_series)

# Re-apply the already-fitted ARIMA(5,1,0) parameters to the full series (no re-estimation, refit=False)
arima_walkforward = arima_fit.apply(full_series, refit=False)

# One-step-ahead predictions across the whole test period:
# dynamic=False means at every hour, the model uses the TRUE previous value, not its own earlier guess
pred = arima_walkforward.get_prediction(start=n_train, end=n_total - 1, dynamic=False)
arima_pred_wf = pred.predicted_mean.values

test_df['ARIMA_Pred_WF'] = arima_pred_wf

# Keep the original blind forecast result for discussion, but label it clearly
results[0]['Model'] = "ARIMA (Blind Multi-Step)"

arima_train_fitted = arima_fit.fittedvalues
results.append(evaluate_forecast(arima_train_series, arima_train_fitted, "ARIMA - Train"))
results.append(evaluate_forecast(test_df['SYSTEM_DEMAND_MW'], test_df['ARIMA_Pred_WF'], "ARIMA (Walk-Forward)"))
```

```python
# ARIMA fold function for cross-validation: fits a fresh (2,1,2) model per fold and forecasts blind, testing how ARIMA generalises across the decade
def arima_fold_fn(train_fold, test_fold):
    order = (2, 1, 2)  # or reuse the order from your auto_arima search
    model = ARIMA(train_fold['SYSTEM_DEMAND_MW'], order=order).fit()
    forecast = model.forecast(steps=len(test_fold))
    return test_fold['SYSTEM_DEMAND_MW'].values, forecast.values

arima_cv_df = run_timeseries_cv(df, arima_fold_fn, label="ARIMA", n_splits=3)
```

```python
# Build lag features (1h, 24h, 168h) since Random Forest has no built-in sense of time
def add_lag_features(data):
    data = data.copy()
    data['Lag_1h']   = data['SYSTEM_DEMAND_MW'].shift(1)
    data['Lag_24h']  = data['SYSTEM_DEMAND_MW'].shift(24)
    data['Lag_168h'] = data['SYSTEM_DEMAND_MW'].shift(168)  # same hour, previous week
    return data

df_lagged = add_lag_features(df)
df_lagged = df_lagged.dropna().reset_index(drop=True)

train_lagged = df_lagged[df_lagged['DateTime'] < SPLIT_DATE].reset_index(drop=True)
test_lagged  = df_lagged[df_lagged['DateTime'] >= SPLIT_DATE].reset_index(drop=True)

CALENDAR_FEATURES = ['Hour', 'DayOfWeek', 'IsWeekend', 'Month', 'Day', 'Year']
LAG_FEATURES      = ['Lag_1h', 'Lag_24h', 'Lag_168h']
TARGET            = 'SYSTEM_DEMAND_MW'

print(f"Train rows: {len(train_lagged):,}  |  Test rows: {len(test_lagged):,}")
```

```python
# Train a baseline Random Forest using calendar features only
from sklearn.ensemble import RandomForestRegressor

X_train_cal = train_lagged[CALENDAR_FEATURES]
y_train_cal = train_lagged[TARGET]
X_test_cal  = test_lagged[CALENDAR_FEATURES]
y_test_cal  = test_lagged[TARGET]

rf_baseline = RandomForestRegressor(n_estimators=200, max_depth=15, random_state=42, n_jobs=-1)
rf_baseline.fit(X_train_cal, y_train_cal)
rf_baseline_pred = rf_baseline.predict(X_test_cal)

rf_baseline_train_pred = rf_baseline.predict(X_train_cal)
results.append(evaluate_forecast(y_train_cal, rf_baseline_train_pred, "RF (Calendar) - Train"))
results.append(evaluate_forecast(y_test_cal, rf_baseline_pred, "RF (Calendar) - Test"))
```

```python
# Train Random Forest with calendar + lag features and compare against the baseline
FEATURES = CALENDAR_FEATURES + LAG_FEATURES

X_train = train_lagged[FEATURES]
y_train = train_lagged[TARGET]
X_test  = test_lagged[FEATURES]
y_test  = test_lagged[TARGET]

rf_full = RandomForestRegressor(n_estimators=300, max_depth=20, random_state=42, n_jobs=-1)
rf_full.fit(X_train, y_train)
rf_full_pred = rf_full.predict(X_test)

rf_full_train_pred = rf_full.predict(X_train)
results.append(evaluate_forecast(y_train, rf_full_train_pred, "RF (Cal+Lag) - Train"))
results.append(evaluate_forecast(y_test, rf_full_pred, "RF (Cal+Lag) - Test"))
test_lagged['RF_Pred'] = rf_full_pred

print("\nComparison: adding lag features to Random Forest")
print(f"  Calendar only : MAE {mean_absolute_error(y_test_cal, rf_baseline_pred):.2f} MW")
print(f"  Calendar+Lag  : MAE {mean_absolute_error(y_test, rf_full_pred):.2f} MW")
```

```python
# Random split comparison for Random Forest
from sklearn.model_selection import train_test_split

train_random, test_random = train_test_split(df_lagged, test_size=0.2, shuffle=True, random_state=42)
train_random = train_random.reset_index(drop=True)
test_random  = test_random.reset_index(drop=True)

X_train_rand = train_random[FEATURES]
y_train_rand = train_random[TARGET]
X_test_rand  = test_random[FEATURES]
y_test_rand  = test_random[TARGET]

rf_random = RandomForestRegressor(n_estimators=300, max_depth=20, random_state=42, n_jobs=-1)
rf_random.fit(X_train_rand, y_train_rand)
rf_random_pred = rf_random.predict(X_test_rand)

results.append(evaluate_forecast(y_test_rand, rf_random_pred, "RF (Cal+Lag) - Random Split"))
```

```python
# Fold function for Random Forest — used by run_timeseries_cv
def rf_fold_fn(train_fold, test_fold):
    X_train, y_train = train_fold[FEATURES], train_fold[TARGET]
    X_test,  y_test  = test_fold[FEATURES],  test_fold[TARGET]
    model = RandomForestRegressor(n_estimators=300, max_depth=20, random_state=42, n_jobs=-1)
    model.fit(X_train, y_train)
    return y_test, model.predict(X_test)

rf_cv_df = run_timeseries_cv(df_lagged, rf_fold_fn, label="Random Forest", n_splits=5)
```

```python
# Scale demand values and build 24-hour sliding-window sequences for the LSTM
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Input
from tensorflow.keras.callbacks import EarlyStopping

WINDOW = 24  # past 24 hours used to predict the next hour

# Fit scaler on TRAIN ONLY to avoid leaking test-set information
scaler = MinMaxScaler()
train_scaled = scaler.fit_transform(train_df[['SYSTEM_DEMAND_MW']])
test_scaled  = scaler.transform(test_df[['SYSTEM_DEMAND_MW']])

def create_sequences(data, window):
    X, y = [], []
    for i in range(window, len(data)):
        X.append(data[i-window:i, 0])
        y.append(data[i, 0])
    return np.array(X), np.array(y)

combined_scaled = np.concatenate([train_scaled[-WINDOW:], test_scaled])

X_train_lstm, y_train_lstm = create_sequences(train_scaled, WINDOW)
X_test_lstm,  y_test_lstm  = create_sequences(combined_scaled, WINDOW)

X_train_lstm = X_train_lstm.reshape((X_train_lstm.shape[0], X_train_lstm.shape[1], 1))
X_test_lstm  = X_test_lstm.reshape((X_test_lstm.shape[0], X_test_lstm.shape[1], 1))

print(f"X_train_lstm: {X_train_lstm.shape}  |  X_test_lstm: {X_test_lstm.shape}")
```

```python
# Define the LSTM model: one LSTM(64) layer feeding into a single output neuron
lstm_model = Sequential([
    Input(shape=(WINDOW, 1)),
    LSTM(64, activation='tanh'),
    Dense(1)
])

lstm_model.compile(optimizer='adam', loss='mse')
lstm_model.summary()
```

```python
# LSTM fold function for cross-validation: fits a fresh scaler and rebuilds 24-hour windows per fold, carrying the last WINDOW training points into test to avoid losing early predictions
def lstm_fold_fn(train_fold, test_fold):
    WINDOW = 24  # same window size as the main LSTM

    # Fit scaler on this fold's TRAIN ONLY - never on test, to avoid leakage
    fold_scaler = MinMaxScaler()
    train_scaled_f = fold_scaler.fit_transform(train_fold[['SYSTEM_DEMAND_MW']])
    test_scaled_f  = fold_scaler.transform(test_fold[['SYSTEM_DEMAND_MW']])

    # Carry the last WINDOW train points into test so the first test predictions
    # have a full 24h of history, same trick as the main LSTM cell
    combined_scaled_f = np.concatenate([train_scaled_f[-WINDOW:], test_scaled_f])

    X_train_f, y_train_f = create_sequences(train_scaled_f, WINDOW)
    X_test_f,  y_test_f  = create_sequences(combined_scaled_f, WINDOW)

    X_train_f = X_train_f.reshape((X_train_f.shape[0], X_train_f.shape[1], 1))
    X_test_f  = X_test_f.reshape((X_test_f.shape[0], X_test_f.shape[1], 1))

    # Fresh small model per fold - keep epochs low, CV already retrains 3x
    fold_model = Sequential([
        Input(shape=(WINDOW, 1)),
        LSTM(64, activation='tanh'),
        Dense(1)
    ])
    fold_model.compile(optimizer='adam', loss='mse')

    early_stop_f = EarlyStopping(monitor='val_loss', patience=3, restore_best_weights=True)
    fold_model.fit(
        X_train_f, y_train_f,
        validation_split=0.1,
        epochs=15,
        batch_size=256,
        callbacks=[early_stop_f],
        verbose=0
    )

    pred_scaled_f = fold_model.predict(X_test_f, verbose=0)
    y_pred = fold_scaler.inverse_transform(pred_scaled_f).flatten()
    y_true = fold_scaler.inverse_transform(y_test_f.reshape(-1, 1)).flatten()

    return y_true, y_pred

lstm_cv_df = run_timeseries_cv(df, lstm_fold_fn, label="LSTM", n_splits=3)
```

```python
# ============================================================
# Build all result tables
# ============================================================
results_df = pd.DataFrame(results)

def classify_set(model_name):
    if 'Random Split' in model_name:
        return 'Random Split'
    elif 'Train' in model_name:
        return 'Train'
    else:
        return 'Test'

results_df['Set'] = results_df['Model'].apply(classify_set)
results_df['Accuracy (%)'] = 100 - results_df['MAPE']

cv_all = pd.concat([rf_cv_df, arima_cv_df, lstm_cv_df], ignore_index=True)
cv_summary = cv_all.groupby('Model').agg(
    MAE_mean=('MAE', 'mean'),   MAE_std=('MAE', 'std'),
    RMSE_mean=('RMSE', 'mean'), RMSE_std=('RMSE', 'std'),
    MAPE_mean=('MAPE', 'mean'), MAPE_std=('MAPE', 'std')
).reset_index().sort_values('RMSE_mean')
cv_summary['Accuracy (%)'] = 100 - cv_summary['MAPE_mean']

# ============================================================
# Print results grouped BY MODEL, with sub-headings per split type
# ============================================================

# Map each model family to the row-label patterns that belong to it
model_groups = {
    "RANDOM FOREST": {
        "train":  ['RF (Cal+Lag) - Train', 'RF (Calendar) - Train'],
        "test":   ['RF (Cal+Lag) - Test', 'RF (Calendar) - Test'],
        "cv":     'Random Forest',
        "random": 'RF (Cal+Lag) - Random Split',
    },
    "LSTM": {
        "train":  ['LSTM - Train'],
        "test":   ['LSTM - Test'],
        "cv":     'LSTM',
        "random": 'LSTM - Random Split',
    },
    "ARIMA": {
        "train":  ['ARIMA - Train'],
        "test":   ['ARIMA (Walk-Forward)', 'ARIMA (Blind Multi-Step)'],
        "cv":     'ARIMA',
        "random": None,  # ARIMA excluded from random-split comparison
    },
}

for model_name, groups in model_groups.items():
    print("=" * 70)
    print(model_name)
    print("=" * 70)

    # --- Train ---
    train_rows = results_df[results_df['Model'].isin(groups["train"])].drop(columns='Set')
    if not train_rows.empty:
        print("\n--- Train (in-sample fit) ---")
        print(train_rows.sort_values('RMSE').to_string(index=False))

    # --- Chronological Test ---
    test_rows = results_df[results_df['Model'].isin(groups["test"])].drop(columns='Set')
    if not test_rows.empty:
        print("\n--- Chronological Test (single 80/20 out-of-sample split) ---")
        print(test_rows.sort_values('RMSE').to_string(index=False))

    # --- TimeSeriesSplit CV ---
    cv_rows = cv_summary[cv_summary['Model'] == groups["cv"]]
    if not cv_rows.empty:
        print("\n--- TimeSeriesSplit Cross-Validation (5-fold chronological) ---")
        print(cv_rows.to_string(index=False))

    # --- Random Split ---
    if groups["random"]:
        rand_rows = results_df[results_df['Model'] == groups["random"]].drop(columns='Set')
        if not rand_rows.empty:
            print("\n--- Random Split (for comparison only — not methodologically valid) ---")
            print(rand_rows.to_string(index=False))

    print()  # blank line between model sections
```
